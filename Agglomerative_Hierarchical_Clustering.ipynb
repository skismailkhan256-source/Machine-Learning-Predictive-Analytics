```python
# Step 1: Load liberis
import matplotlib.pyplot as plt
import pandas as pd

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from sklearn.decomposition import PCA

from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
# Step 2: Load the dataset
iris = load_iris()
X = iris.data

# Select the first 50 data points
X = X[:50]

print("Dataset shape:", X.shape)
print(X)
```

```python
# Step 3: Standardize the Data
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Standardized data:")
print(X_scaled)
```

```python
# Step 4: Calculate the Linkage Matrix
linked = linkage(X_scaled, method='ward')
print("Linkage matrix:")
print(linked)
```

```python
# Step 5: Draw the Dendrogram
plt.figure(figsize=(12, 6))

dendrogram(
    linked,
    labels=list(range(1, len(X) + 1)),
    leaf_rotation=90,
    leaf_font_size=8
)

plt.title("Dendrogram - Agglomerative Hierarchical Clustering")
plt.xlabel("Data Point")
plt.ylabel("Euclidean Distance")
plt.tight_layout()
plt.show()
```

```python
# Step 6: Create the clustering model

model = AgglomerativeClustering(
    n_clusters=3,
    linkage='ward'
)
```

```python
# Step 7: Perform Agglomerative Clustering

labels = model.fit_predict(X_scaled)

print("Cluster labels:")
print(labels)
```

```python
# Step 8: Display the cluster assignments

results = pd.DataFrame({
    "Data Point": range(1, len(X) + 1),
    "Cluster": labels
})

print(results.to_string(index=False))
```

```python
# Step 9: Reduce the data to two dimensions

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Reduced data shape:", X_pca.shape)
```

```python
# Step 10: Visualize the clusters

plt.figure(figsize=(9, 6))

scatter = plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=labels,
    cmap="viridis",
    s=80,
    edgecolors="black"
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Agglomerative Hierarchical Clustering")
plt.colorbar(scatter, label="Cluster")
plt.tight_layout()
plt.show()
```
