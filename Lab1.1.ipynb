# Lab_1 :Clean a raw dataset and handle missing values, Perform feature enginering on a dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Creat a data set
data = {
    "Name": [
        "Amit", "Priya", "Rahul", "Sneha", "Rohan",
        "Anjali", "Vikash", "Neha", "Suman", "Karan",
        "Pooja", "Arjun", "Meena", "Rakesh", "Nisha",
        "Deepak", "Kavita", "Ajay", "Simran", "Ravi",
        "Rahul"
    ],

    "Age": [
        25, 30, np.nan, 28, 35,
        22, 29, np.nan, 31, 27,
        26, 34, 24, 32, np.nan,
        29, 23, 36, 28, 30,
        np.nan
    ],

    "Gender": [
        "Male","Female","Male","Female","Male",
        "Female","Male","Female","Female","Male",
        "Female","Male","Female","Male","Female",
        "Male","Female","Male","Female","Male",
        "Male"
    ],

    "Department": [
        "HR","IT","Sales","Finance","IT",
        "HR","Sales","Finance","IT","HR",
        "Sales","IT","Finance","HR","Sales",
        "IT","Finance","HR","Sales","IT",
        "Sales"
    ],

    "Salary": [
        40000,50000,45000,np.nan,60000,
        38000,52000,48000,np.nan,47000,
        43000,62000,39000,58000,45000,
        np.nan,41000,65000,49000,53000,
        45000
    ],

    "Experience": [
        2,5,3,4,np.nan,
        1,6,2,5,3,
        np.nan,7,2,6,4,
        5,1,8,np.nan,4,
        3
    ],

    "Performance": [
        "Good","Excellent","Average","Good","Excellent",
        "Average","Good","Excellent","Good","Average",
        "Good","Excellent","Average","Good","Excellent",
        "Good","Average","Excellent","Good","Average",
        "Average"
    ]
}

df = pd.DataFrame(data)
```

```python
df.tail()
```

```python
df.isnull().sum()
```

```python
sns.heatmap(df.isnull(), cmap="viridis")
plt.show()
```

```python
df["Age"]=df["Age"].fillna()(df["Age"].mean())
df["Salary"]=df["Salary"].fillna()(df["Salary"].mode()[0])
df["Experience"]=df["Experienece"].fillna()(df["Experience"])
```

```python
df.isnull().sum()
```

# 2.Feature enginering

```python
# Label encoding
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df["Gender"] = le.fit_transform(df["Gender"])
df["Department"] = le.fit_transform(df["Department"])
```

```python
# performm a Feature Engineering
df["Annual_Salary"] = df["Salary"]*12 
```

```python
df["Senior"] = df["Experience"] >=5
```
